# Pandas Fundamentals

A beginner-friendly introduction to Series, DataFrames, selection, filtering, missing values, and summary statistics.

```python
import numpy as np
import pandas as pd

print("Numpy version:", np.__version__)
print("Pandas version:", pd.__version__)
```

```python
#create a padas series of student marks
marks = pd.Series([85, 55, 90, 58, 75,56])

print(marks)
```

```python
#accessing the elements of the series using index
import pandas as pd
marks = pd.Series([85, 55, 90, 58, 75, 56], index=['John', 'Alice', 'Bob', 'Eve', 'Charlie', 'David'])
print(marks)


print(marks['John'])

display(marks['John'])


print(marks[['Alice', 'Bob']])

print("highest marks:", marks.max())

print("lowest marks:", marks.min())

print("average marks:", marks.mean())
```

```python
#dataframe creation
students={
    "name":["john", "Alice", "Bob", "Eve", "Charlie", "David"],
    "branch":["CSE", "ECE", "MECH", "CSE", "ECE", "MECH"],
    "marks":[85, 55, 90, 58, 75, 56],
    "attendance":[90, 80, 95, 85, 70, 75]
}

df=pd.DataFrame(students)

display(df)

display(df.tail(2))
```

```python
#displaying specific columns
display(df[["name","marks"]])
```

```python
#access using iloc
print(df.iloc[0])

#access using integer location
print(df.iloc[1])

#access the sepcific row and column (integer position)
print(df.iloc[1,3])

print("loc\n")
print(df.loc[0:2])

#based on index label
print(df.loc[0,"marks"])
```

```python
#filter students
students={
    "name": ["rahul","priya","arjun","sneha","kiran"],
    "branch": ["CSE","AIML","ESE","CSE","IT"],
    "marks": [85,90,75,95,88]
}

df=pd.DataFrame(students)

display(df)
```

```python
high_score = df[df["marks"] > 80]

display(high_score)
```

```python
#using &(and) |(or) ~ (not)
result = df [(df["marks"]>80)&(df["branch"]=="CSE")]

display(result)
```

```python
#sorting
display(df.sort_values(by="marks"))

display(df.sort_values(by ="marks",ascending=False))
```

```python
#sorting multiple columns
display(
    df.sort_values(
        by=["branch", "marks"],
        ascending=[True, False]
    )
)
```
