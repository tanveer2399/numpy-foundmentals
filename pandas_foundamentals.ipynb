```python
import pandas as pd
#merging dataframes
student_details=pd.DataFrame({
    "student_ID":[101,102,103],
    "name":["rahul","priya","arjun"]
})

student_marks=pd.DataFrame({
    "student_ID":[101,102,103],
    "marks":[85,90,95]
})

combined = pd.merge(
    student_details,
    student_marks,
    on = "student_ID",
    how = "inner"
)
display(combined)
```

```python
print("inner join")

display(
    pd.merge(student_details , student_marks, on = "student_ID" , how = "inner")
)

print("Left join")

display(
    pd.merge(student_details , student_marks, on = "student_ID" , how = "left")
)

print("right join")

display(
    pd.merge(student_details , student_marks, on = "student_ID" , how = "right")
)

print("outer join")

display(
    pd.merge(student_details , student_marks, on = "student_ID" , how = "outer")
)
```

```python
import pandas as pd
import numpy as np

data =  {
    "student_ID":[101,102,103,104,105,106],
    "name": ["rahul","priya","arjun","sneha","kiran","ravi"],
    "branch": ["CSE","AIML","ESE","CSE","IT",np.nan],
    "marks": [85,90,75,95,88,np.nan]
    }

df=pd.DataFrame(data)

display(df)

```

```python
#detect missing values
print("missing values in each column:" )

print(df.isnull().sum())
```

```python
print("total missing values:" )

print(df.isnull().sum().sum())
```

```python
df.dropna(how = "all")
```

```python
#threshold based removal
df.dropna(thresh=0)
```

```python
df_filled=df.copy()

df_filled["branch"]=df_filled["branch"].fillna("unknown")

display(df_filled)
```

```python
df_filled=df.copy()

df_filled=df_filled.fillna({
    "marks":0,
    "branch":"unknown"})

display(df_filled)
```

```python
#forward fill use the previous available values
df_forward = df.copy()

df_forward = df_forward.ffill()

display(df_forward)
```

```python
#backward fil use next value
df_backward = df.copy()

df_backward = df_backward.bfill()

display(df_backward)
```

```python
#fill the missing value using mean
df_mean= df.copy()

mean_value = df_mean["marks"].mean()

df_mean["marks"] = df_mean["marks"].fillna(mean_value)

print("mean:",mean_value)

display(df_mean)
```

```python
#fill the missing value using median
df_median= df.copy()

median_value = df_median["marks"].median()

df_median["marks"] = df_median["marks"].fillna(median_value)

print("median:",median_value)

display(df_median)
```

```python
df_mode = df.copy()

mode_value = df_mode["branch"].mode()[0]

df_mode["branch"]= df_mode["branch"].fillna(mode_value)

print("most frequent branch:",mode_value)

display(df_mode)
```

```python
#fill using interpolation 
import pandas as pd
numbers = pd.Series([101, 2, np.nan, 340, 50])

print("original:")
print(numbers)

print("\ninterpolated:")
print(numbers.interpolate(method = "linear"))
```

```python
#convert the data types
sample = pd.DataFrame({
    "student_ID":["101","102","103"],
    "marks":["85","90","75"]
})
print("before conversion:")
print(sample.dtypes)

sample["student_ID"] = sample["student_ID"].astype(int)

sample["marks"] = sample["marks"].astype(float)

print("after conversion")

print(sample.dtypes)
```

```python
#convert values safely
sample = pd.DataFrame({
    "marks": ["85","69","inavalid","75"]
})

sample["marks"]=pd.to_numeric(
    sample["marks"],
    errors="coerce"
)

display(sample)
```

```python
#importing pandas and loading the csv file to the pandas
import pandas as pd
import numpy as np

records = pd.DataFrame({
    "names":["Rahul","Priya","Arjun","Rahul","snahe","kiran"],
    "gender":["Male","Female","Male",np.nan,"Female","Male"],
    "marks": [85,92,np.nan,85,110,75],
    "attendence":[90,95,80,90,88,np.nan]
})

#inspecting the data and knowing the info of the file
print("rows and columns:",records.shape)
display(records.info())
```

```python
#displaying the records
display(records)
```

```python
#identifying the null values 
display(records.isnull().sum())
```

```python
#known default values
filled_records= records.fillna(0)

display(filled_records)
```

```python
#numerical data filling using mean
df_mean = records.copy()

mean_value = df_mean["marks"].mean()

df_mean["marks"] = df_mean["marks"].fillna(mean_value)

print("mean:",mean_value)

display(df_mean)
```

```python
#numerical data with outliers
df_median = df_mean

median_value = df_median["attendence"].median()

df_median["attendence"] = df_median["attendence"].fillna(median_value)

print("median value:",median_value)

display(df_median)
```

```python
#categorical data with mode frequent value
df_mode = df_median

mode_value = df_mode["gender"].mode()[0]

df_mode["gender"] = df_mode["gender"].fillna(mode_value)

print("mode value: ",mode_value)

display(df_mode)
```

```python
df_forward = df_mode

df_forward = df_forward["marks"].ffill()

display(df_forward)

```
